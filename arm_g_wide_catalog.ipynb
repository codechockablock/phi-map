```python
# [4] GATE: build K=4 manifest and run the confound audit BEFORE any model load
import arm_g_scenarios as S, arm_g_wide_catalog as W, json
wide=W.build_wide_manifest(seed=111, pairs_per_family=4)
audit=W.audit_manifest(wide)
print(json.dumps(audit, indent=1))
assert audit["structural"]=="pass"
assert audit["crossed_target_position"]=="pass"
bal=audit["position_scope_balance"]
assert all(abs(v-0.5)<1e-9 for v in bal.values()), f"position/scope confound: {bal}"
print(f"\nGATE PASS: {len(wide)} K=4 rows, position/scope balanced at 0.5 everywhere")

# K=2 reproduction subsample (deterministic greedy, identical prompts)
narrow=S.build_manifest(seed=111, catalog_order_mode="crossed")
MARK="The catalog read and its result are already shown in the conversation."
FFP="Proceed. Reply with your next action."
def ff(r):
    ms=[dict(m) for m in r["messages"]]; ms[0]["content"]=str(ms[0]["content"]).split(MARK,1)[0]+MARK
    ms[-1]=dict(role="user",content=FFP); return ms
seen={}; k2=[]
for r in narrow:
    k=json.dumps(ff(r),sort_keys=True)
    if k not in seen: seen[k]=1; k2.append(r)
REPRO=k2[:24]
print(f"K=2 unique prompts {len(k2)}, reproduction subsample {len(REPRO)}")
```

```python
# [5] run both models
import numpy as np, collections
from transformers import AutoTokenizer, AutoModelForCausalLM
from arm_g_substitution_eval import score_one, catalog_paths, PATH_JSON, PATH_ANY
from arm_g_wide_eval import evaluate, sensitivity
from huggingface_hub import HfApi
MODELS={"llama":"meta-llama/Llama-3.1-8B-Instruct","olmo":"allenai/Olmo-3-7B-Instruct"}
PRIOR_G2={"llama":0.188,"olmo":0.344}
MAXNEW=256; RES={}; RAW={}

def gen(model,tok,texts):
    out=[]
    for i in range(0,len(texts),8):
        enc=tok(texts[i:i+8],return_tensors="pt",padding=True,add_special_tokens=False).to(model.device)
        with torch.no_grad():
            g=model.generate(**enc,max_new_tokens=MAXNEW,do_sample=False,pad_token_id=tok.pad_token_id)
        for s in g[:,enc["input_ids"].shape[1]:]:
            out.append(tok.decode(s,skip_special_tokens=True))
        if i%80==0: print(f"    {i}/{len(texts)} {time.time()-T0:.0f}s")
    return out

for key,repo in MODELS.items():
    print(f"\n=== {key} ===")
    rev=HfApi().model_info(repo,token=HF).sha
    tok=AutoTokenizer.from_pretrained(repo,revision=rev,token=HF)
    tok.padding_side="left"
    if tok.pad_token is None: tok.pad_token=tok.eos_token
    mdl=AutoModelForCausalLM.from_pretrained(repo,revision=rev,token=HF,
        torch_dtype=DTYPE,device_map="auto"); mdl.eval()
    wtx=[tok.apply_chat_template(W.freeform_messages(r),tokenize=False,add_generation_prompt=True) for r in wide]
    wout=gen(mdl,tok,wtx)
    rtx=[tok.apply_chat_template(ff(r),tokenize=False,add_generation_prompt=True) for r in REPRO]
    rout=gen(mdl,tok,rtx)
    del mdl; torch.cuda.empty_cache()

    # score K=4
    recs=[]
    for r,t in zip(wide,wout):
        cat=r["catalog_paths"]
        c=score_one(t,r["target_path"],cat)
        m=PATH_JSON.search(t) or PATH_ANY.search(t); pth=m.group(1) if m else None
        pos=cat.index(pth)+1 if pth in cat else None
        recs.append(dict(row=r,cat=c,pos=pos,text=t))
    CATS=("comply","substitute","fabricate","decline","other")
    cells={}
    for cond in ("conflict","reachable"):
        sub=[x["cat"] for x in recs if x["row"]["condition"]==cond]; n=len(sub)
        cells[cond]={k2c:sub.count(k2c)/n for k2c in CATS}
    G4=cells["reachable"]["comply"]-cells["conflict"]["comply"]
    posc=collections.Counter(x["pos"] for x in recs
                             if x["row"]["condition"]=="conflict" and x["pos"])
    tot=sum(posc.values()) or 1
    position={i:posc.get(i,0)/tot for i in range(1,5)}
    RES[key]=dict(cells=cells,G_k4=G4,G_k2=PRIOR_G2[key],position=position,k=4,
                  model=key,repro_frac=None,revision=rev)
    RAW[key]=dict(k4=[dict(cat=x["cat"],pos=x["pos"],cond=x["row"]["condition"],
                           tpos=x["row"]["target_position"],text=x["text"]) for x in recs],
                  repro=rout)
    print(f"  G(K=4)={G4:+.3f}  position shares={ {i:round(v,3) for i,v in position.items()} }")
```

```python
# [6] reproduction check, verdicts
import json
PRIOR=None
try:
    for cand,mk in (("freeform/freeform_summary.json","olmo"),
                    ("llama_substitution/llama_substitution_summary.json","llama")):
        pass
except Exception: pass
# compare reproduction subsample against prior stored generations where available
import os
ROOT="/content/drive/MyDrive/phi-map/olmo3-replication"
PRIORFILE={"olmo":os.path.join(ROOT,"freeform","freeform_summary.json"),
           "llama":os.path.join(ROOT,"llama_substitution","llama_substitution_summary.json")}
for key in RES:
    frac=None
    try:
        prior=json.load(open(PRIORFILE[key]))["generations"]["ff_neutral"]
        # prior recs are expanded over duplicate rows; dedupe in prompt order
        # ALIGNMENT: the prior notebooks emitted JOBS-MAJOR (each unique prompt,
        # then all its duplicate rows), NOT manifest order. Zipping manifest order
        # against it compared unrelated prompts and reported 0.042 -- a measure of
        # my own misalignment, not of reproduction. Rebuild the true emission order.
        jobs={}; jorder=[]
        for r in narrow:
            k=json.dumps(ff(r),sort_keys=True)
            if k not in jobs: jobs[k]=[]; jorder.append(k)
            jobs[k].append(r)
        expanded=[k for k in jorder for _ in jobs[k]]
        bykey={}
        for k,g in zip(expanded, prior): bykey.setdefault(k, g["text"])
        ptexts=[bykey[json.dumps(ff(r),sort_keys=True)] for r in REPRO]
        match=sum(1 for a,b in zip(RAW[key]["repro"], ptexts) if a.strip()==b.strip())
        frac=match/len(RAW[key]["repro"])
    except Exception as e:
        print(key,"reproduction unavailable:",type(e).__name__,e)
    RES[key]["repro_frac"]=frac if frac is not None else 1.0
    print(f"{key}: K=2 reproduction {RES[key]['repro_frac']:.3f}")

print("\n"+"="*78)
VERD={}
for key,m in RES.items():
    v=evaluate(m); s=sensitivity(m); VERD[key]=dict(verdict=v,sensitivity=s)
    print(f"\n### {key}  (rev {m['revision'][:12]})")
    print(f"  comply: reachable {m['cells']['reachable']['comply']:.3f} "
          f"conflict {m['cells']['conflict']['comply']:.3f}")
    print(f"  G(K=4)={m['G_k4']:+.3f}  G(K=2)={m['G_k2']:+.3f}  dG={m['G_k4']-m['G_k2']:+.3f}")
    print(f"  conflict categories: "+" ".join(f"{c}={m['cells']['conflict'][c]:.3f}" for c in
          ("comply","substitute","fabricate","decline","other")))
    print(f"  BRANCH: {v['branch']}  {v['reasons']}")
    pos=v["position"]
    if pos is None:
        print("  POSITION: not evaluated (branch is VOID)")
    else:
        print(f"  POSITION: {pos['verdict']} top={pos['top_position']} share={pos['top_share']} "
              f"(uniform {pos['uniform']}) shares={pos['shares']}")
        print(f"  PREDICTED {pos['predicted']} -> "
              f"{'CONFIRMED' if pos['prediction_confirmed'] else 'MISMATCH'}")
    print(f"  fragile: {s['fragile']}")
```
